# Product Engineering Lab · Nyfasi research reconstruction

This notebook builds an **illustrative** component-level proxy for the GB2591502A hair-applicator architecture. Its direct relationship to the Nyfasi-branded Deluxe Detangler remains unverified. It is not official Nyfasi CAD, a verified retail replica or manufacturing geometry.

Sand3D documents Python 3.11 or 3.13. The notebook uses uv to provision Python 3.13 even when the hosted notebook kernel itself uses another version.

```python
import sys, subprocess, shutil
print('Notebook kernel:', sys.version)
subprocess.run([sys.executable, '-m', 'pip', 'install', 'uv'], check=True)
uv = shutil.which('uv')
if not uv:
    raise RuntimeError('uv CLI was not found after installation.')
subprocess.run([uv, 'python', 'install', '3.13'], check=True)
print('Provisioned a Sand3D-supported Python 3.13 runtime.')
```

```python
import subprocess
subprocess.run(['git', 'clone', 'https://github.com/andiekobbietks/product-engineering-lab.git'], check=True)
print('Repository cloned.')
```

```python
from pathlib import Path
import subprocess
repo = Path('/content/product-engineering-lab')
script = repo / 'experiments/001-nyfasi-reconstruction/scripts/build_reconstruction.py'
subprocess.run([uv, 'run', '--python', '3.13', '--with', 'sand3d==1.6.2', 'python', str(script)], cwd=repo, check=True)
print('Exports:', list((repo / 'experiments/001-nyfasi-reconstruction/output').glob('*')))
from google.colab import files
files.download(str(repo / 'experiments/001-nyfasi-reconstruction/output' / 'hair_applicator_research_exploded.glb'))
```
